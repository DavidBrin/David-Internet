## save sweeps as *pickle* files

```python
import pickle
```

function to extract data, copied from download and parameterization notebooks

```python
def extract_data(file_path):
    # Open the HDF5 file
    with h5py.File(file_path, 'r') as f:
        # Initialize an empty dictionary to store data arrays
        data = {}

        # Iterate through keys in the 'acquisition' group
        for sweep_key in f['acquisition'].keys():
            dataset = f['acquisition'][sweep_key]['data'] 
            # Convert the dataset data into a NumPy array and store in the dictionary
            data[sweep_key] = np.array(dataset)

    return data
```

```python
for file in filepaths:
    data = extract_data(file)
    # Extract filename without extension
    filename_prefix = os.path.splitext(os.path.basename(file))[0]  
    for key, value in data.items():
        with open(f'{filename_prefix}_{key}.pkl', 'wb') as f:
            pickle.dump(value, f)
```
